```python
import cadquery as cq
from cadquery import exporters
import math
from fractions import Fraction
from jupyter_cadquery import *

from jupyter_cadquery.replay import replay, enable_replay, disable_replay

import warnings
warnings.filterwarnings("ignore", category=DeprecationWarning)

versions()
```

```python
cv = open_viewer("Post", cad_width=800, height=600)
set_default_viewer("Post")
```

```python
# ------------------------------------------------------------
# SILL
# ------------------------------------------------------------

sill_width = 8.0
sill_height = 10.0

sill_x_length = 72.0
sill_y_length = 72.0

# Half lap
lap_depth = sill_height / 2


# ------------------------------------------------------------
# POST
# ------------------------------------------------------------

post_width = 6.0
post_depth = 4.0
post_height = 96.0

# Bottom tenon
tenon_width = 3.0       # X
tenon_depth = 2.0       # Y
tenon_length = (sill_height/2)      # Z, into sill

mortise_clearance = 0.02
mortise_extra_depth = 0.25
```

```python
sill_x = cq.Solid.makeBox(
    sill_x_length,
    sill_width,
    sill_height,
    cq.Vector(
        0,
        0,
        0,
    ),
)
```

```python
sill_y = cq.Solid.makeBox(
    sill_width,
    sill_y_length,
    sill_height,
    cq.Vector(
        0,
        0,
        0,
    ),
)
```

```python
x_lap_cutter = cq.Solid.makeBox(
    sill_width,
    sill_width,
    lap_depth,
    cq.Vector(
        0,
        0,
        sill_height - lap_depth,
    ),
)

sill_x_lapped = (
    sill_x
    .cut(x_lap_cutter)
    .clean()
)
```

```python
y_lap_cutter = cq.Solid.makeBox(
    sill_width,
    sill_width,
    lap_depth,
    cq.Vector(
        0,
        0,
        0,
    ),
)

sill_y_lapped = (
    sill_y
    .cut(y_lap_cutter)
    .clean()
)
```

```python
post_x0 = 0.0
post_y0 = 0.0
```

```python
post_body = cq.Solid.makeBox(
    post_width,
    post_depth,
    post_height,
    cq.Vector(
        post_x0,
        post_y0,
        sill_height,
    ),
)
```

```python
tenon_x0 = (
    post_x0
    + (post_width - tenon_width) / 2
)

tenon_y0 = (
    post_y0
    + (post_depth - tenon_depth) / 2
)
```

```python
post_tenon = cq.Solid.makeBox(
    tenon_width,
    tenon_depth,
    tenon_length,
    cq.Vector(
        tenon_x0,
        tenon_y0,
        sill_height - tenon_length,
    ),
)
```

```python
post = (
    post_body
    .fuse(post_tenon)
    .clean()
)
```

```python
mortise_width = (
    tenon_width + mortise_clearance
)

mortise_depth = (
    tenon_depth + mortise_clearance
)

mortise_length = (
    tenon_length
    + mortise_extra_depth
)
```

```python
mortise_x0 = (
    tenon_x0 - mortise_clearance / 2
)

mortise_y0 = (
    tenon_y0 - mortise_clearance / 2
)
```

```python
post_mortise = cq.Solid.makeBox(
    mortise_width,
    mortise_depth,
    mortise_length,
    cq.Vector(
        mortise_x0,
        mortise_y0,
        sill_height - mortise_length,
    ),
)
```

```python
sill_y_cut_lapped = (
    sill_y_lapped
    .cut(post_mortise)
    .clean()
)
```

```python
show_object(sill_y_cut_lapped, name="sill_y")
show_object(sill_x_lapped, name="sill_x")
show_object(post, name="post")
```
